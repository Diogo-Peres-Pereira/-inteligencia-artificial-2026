**Exercicio 1**

 integrate.quad(f, 0, 3.14): Calcula a integral definida da função f(x) = x
2
sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy import integrate

def f(x):
  return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, 3.14)
print("O resultado da integração é:", resultado)
print("O erro da integração é:", erro)

```

**Exercicio 2**

solve ivp: Resolve equações diferenciais ordinárias (EDOs) com condições
iniciais.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(x, y):
  return x - y

solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 100))
print("Tempo: ", solucao.t)
print("Solução pra Y: ", solucao.y[0])
```

**Exercicio 3**

root(f, 2): Encontra a raiz da funçãoo f(x) = 0 partindo de uma aproximação inicial x = 2.

```python
from scipy.optimize import root
def f(x):
  return x**4 - 3*x**3 + 2
solucao = root(f, 2)
print("Raiz encontrada: ", solucao.x)
```

**Exercicio 4**

minimize(f, 0): Busca o valor de x que minimiza a funçõa, partindo de
uma estimativa inicial x = 0.
Como não tem maximiza então tem que inverter o sinal da função

```python
from scipy.optimize import minimize
def f(x):
  return -x**2 + 4*x + 1
def f_negativo(x):
  return -f(x)
solucao = minimize(f_negativo, 0)
print("Máximo encontrado: ", solucao.x)
```

**Exercicio 5**

 fft(sinal): Calcula a transformada de Fourier do sinal f(t).

**Exercicio 9**

curve fit: Ajusta a curva com base nos dados fornecidos e na função
modelo. Aqui ajustamos uma parábola.


```python
from scipy.optimize import curve_fit
import numpy as np
matriz = [[1,2], [2,4], [3,5], [4,7], [5,8]]
matriz_np = np.array(matriz)

def curva_quadrada(x, a, b, c):
    return a*x**2 + b*x + c

parametros, _ = curve_fit(curva_quadrada, matriz_np[:,0], matriz_np[:,1])
print("Parâmetros da curva quadrática: ", parametros)
```

**Exercicio 16**

Use a mesma recla do exercio 2, só que antes transformar a equação de segunda ordem num sistema de duas equações de primeira ordem.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(t, y):
  y0, y1 =y
  dy0_dt = y1
  dy1_dt = -y0
  return [dy0_dt, dy1_dt]
avaliacao = np.linspace(0, 10, 100)
solucao = solve_ivp(eq_diferencial, [0, 10], [1, 0], t_eval=avaliacao)
print("Tempo:", solucao.t)
print("Solução para y0:", solucao.y[0])
print("Solução para y1:", solucao.y[1])

```
