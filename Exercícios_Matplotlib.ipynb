**Exercicio 1**

np . linspace cria um grafico de linha e plt.plot(x, y): Cria um gráfico de linha com os valores de x e y

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 100, 10)
y = x**3
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Exercício')
plt.show()
```

**Exercicio 5**

Crie um grafico de pizza, mostrando as porcentagens

```python
import matplotlib.pyplot as plt

labels = ["Setor 1", "Setor 2", "Setor 3", "Setor 4"]
sizes = [15, 30, 45, 50]
colors = ['red', 'green', 'blue', 'purple']

plt.pie(sizes, labels=labels, colors=colors, autopct="%1.1f%%", startangle=90)
plt.title("5 - Gráfico de Pizza")
plt.show()
```

**Exercicio 6**

 plt.tight layout(): Ajusta automaticamente o layout dos gráficos para
evitar sobreposição de títulos e eixos.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 10, 100 )

y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("6 - Gráfico 1")

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("6 - Gráfico 2")

plt.tight_layout()
plt.show()
```

**Exercicio 8**

plt.fill_between(): esse é o comando que faz o sombreamento

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
margem = 0.2
plt.figure()
plt.plot(x, y)
plt.fill_between(x, y + margem, y - margem, color='blue', alpha=0.2)
plt.title("8 - Gráfico de Linha com Área Preenchida")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

**Exercicio 10**

desenhar as linhas antes do plt.show() para aparecerem em um unico grafico

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure()
plt.plot(x, y1, linestyle="--", label='Seno')
plt.plot(x, y2, linestyle=":",label='Cosseno')
plt.plot(x, y3, linestyle="-.",label='sin * cos')
plt.legend()
plt.title("10 - três gráficos")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True, alpha=0.3)
plt.show()

```

**Exercicio 13**

projection='3d': ativa o modo 3D dos eixos

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z)
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_zlabel('Eixo Z')
plt.title("13 - Gráfico de Dispersão 3D")
plt.show()
```

**Exercicio 18**

faz uma condição para separa as cores

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C', 'D', 'E', 'F']
valores = [10, -5, 8, -12, 15, -3]

cores = ['green' if v >= 0 else 'red' for v in valores]

plt.figure()
plt.bar(categorias, valores, color=cores)

plt.title("18 - Gráfico de Barras com Valores Positivos e Negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.axhline(0, color='black', linewidth=0.8)
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

**Exercicio 20**

plt.xticks() / plt.yticks() controla os espaçamentos

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = x**2

plt.figure()
plt.plot(x, y)
plt.title("20 - Grade persolalizada")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.xticks(np.arange(0, 11, 1))
plt.yticks(np.arange(0, 101, 10))
plt.grid(True, linestyle='-.', linewidth=0.7, color='#000080', alpha=0.5)
plt.show()
```
