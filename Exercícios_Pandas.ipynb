**Exercicio 8**

```python
import pandas as pd

dados = {'Produtos': ['Arroz', 'Feijão', 'Porco', 'Macarrão', 'Batata'],
         'Categoria': ['Grãos', 'Grãos', 'Carnes', 'Grãos', 'Vegetal'],
         'Preços': [20.00, 16.00, 38.00, 15.00, 9.00],
         'Quantidade': [50, 100, 30, 100, 300]}

df = pd.DataFrame(dados)
df = (df.groupby("Categoria")["Quantidade"].sum())

print(df)
```

**Exercicio 11**

**Exercicio 13**

df.pivot_table() cria uma tabela dinâmica .

```python
import pandas as pd

dados = {'Produtos': ['Arroz', 'Feijão', 'Porco', 'Macarrão', 'Batata'],
         'Categoria': ['Grãos', 'Grãos', 'Carnes', 'Grãos', 'Vegetal'],
         'Preços': [20.00, 16.00, 38.00, 15.00, 9.00],
         'Quantidade': [50, 100, 30, 100, 300]}

df = pd.DataFrame(dados)
tabela_pivo = df.pivot_table(
    values="Quantidade",
    index="Categoria",
    aggfunc="sum"
)

print(tabela_pivo)
```

**Exercicio 14**

pd.merge junta os dois DataFrames combinando as linhas onde o valor de ID é igual nos dois.

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Produto': ['Arroz', 'Feijão', 'Carne', 'Batata']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Preço': [20.00, 16.00, 38.00, 12.00]
})

df_merged = pd.merge(df1, df2, on='ID', how='inner')
print(df_merged)
```

**Exercicio 15**

.interpolate(method='linear') preenche os NaN traçando uma "linha reta" entre o valor anterior e o próximo valor conhecido, distribuindo os passos igualmente.

```python
import pandas as pd
import numpy as np

dados = {'Produtos': ['Arroz', 'Feijão', 'Carne', 'Macarrão', 'Batata'],
         'Preços': [ 20.00, None, 38.00, None, 9.00]}

df = pd.DataFrame(dados)
df['Preços'] = df['Preços'].interpolate(method='linear')
print(df)

```
