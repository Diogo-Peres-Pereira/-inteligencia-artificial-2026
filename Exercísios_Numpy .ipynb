**Exercicio 7**

Cria os dois array depois pega o coeficiente de correlação

```python
import numpy as np
array1 = np.random.randint(1,10,10)
array2 = np.random.randint(1,10,10)
correlacao = np.corrcoef(array1, array2)
correlacao = np.round(correlacao,2)
valor_correlacao = correlacao[0, 1]
print("Array 1:", array1)
print("Array 2:", array2)
print("Matriz completa:\n", correlacao)
print(f"Correlação entre os arrays: {valor_correlacao}")
```

**Exercicio 11**

Gera 100 numeros aleatorios, use np.unique para retorna dois arrays, um com valores unicos e outro com quantas vezes cada valor único aparece
depois ver qual se repete mais

```python
import numpy as np
array = np.random.randint(1, 100, size=100)
valores_unicos, contagem = np.unique(array, return_counts=True)
indice = np.argmax(contagem)
valor_mais_frequente = valores_unicos[indice]
quantidade = contagem[indice]
print(array)
print("numero que mais se repete ", valor_mais_frequente)
print("quantidade de vezes que se repete", quantidade)

```

**Exercicio 19**

Usar as diferentes funçõs para encontra o cosseno e inverter em radianos depois converter para graus

```python
import numpy as np
array1 = np.random.randint(1,101, size=20)
array2 = np.random.randint(1,101, size=20)
pridutorio = np.dot(array1, array2)
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)
angulo_rad = np.arccos(pridutorio / (norma1 * norma2))
angulo_graus = np.degrees(angulo_rad)
print(array1)
print(array2)
print(pridutorio)
print(angulo_rad)
print(angulo_graus)
```
